```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv("battery_profile.csv")
df1["Timestamp"] = pd.to_datetime(df1["Timestamp"])

random_date = np.random.choice(df1["Timestamp"].dt.date.unique())
one_day_df1 = df1[df1["Timestamp"].dt.date == random_date]

# Plot Battery Charge/Discharge
plt.figure(figsize=(10,5))
plt.plot(one_day_df1["Timestamp"], one_day_df1["Battery_Charge_kW"], label="Charge (kW)")
plt.plot(one_day_df1["Timestamp"], one_day_df1["Battery_Discharge_kW"], label="Discharge (kW)")
plt.title(f"Battery Charging/Discharging – {random_date}")
plt.xlabel("Time"); plt.ylabel("Power (kW)"); plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 1. Battery Charging and Discharging Profile
plt.figure(figsize=(12,5))
plt.plot(df1["Timestamp"], df1["Battery_Charge_kW"], label="Battery Charge")
plt.plot(df1["Timestamp"], df1["Battery_Discharge_kW"], label="Battery Discharge")
plt.title("Battery Charging and Discharging Profile")
plt.xlabel("Time")
plt.ylabel("Power (kW)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2. State of Charge (SOC)

random_date = np.random.choice(df1["Timestamp"].dt.date.unique())
one_day_df1 = df1[df1["Timestamp"].dt.date == random_date]


plt.figure(figsize=(10,5))
plt.plot(one_day_df1["Timestamp"], one_day_df1["SOC_kWh"], label="SOC", color='green')
plt.title(f"State of Charge – {random_date}")
plt.xlabel("Time"); plt.ylabel("SOC (kWh)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,5))
plt.plot(df1["Timestamp"], df1["SOC_kWh"], color='green')
plt.title("Battery State of Charge (SOC) Over Time")
plt.xlabel("Time")
plt.ylabel("SOC (kWh)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df2 = pd.read_csv("optimal_hourly_flow.csv")
df2["Timestamp"] = pd.to_datetime(df2["Timestamp"])

random_date = np.random.choice(df2["Timestamp"].dt.date.unique())
one_day2 = df2[df2["Timestamp"].dt.date == random_date]


plt.figure(figsize=(12,6))
for col in ["Load_kW", "Solar_kW", "Battery_Charge_kW", "Battery_Discharge_kW", "Grid_Import_kW", "Grid_Export_kW"]:
    plt.plot(one_day2["Timestamp"], one_day2[col], label=col)
plt.title(f"Power Flow Components – {random_date}")
plt.xlabel("Time"); plt.ylabel("Power (kW)"); plt.legend()
plt.xticks(rotation=45); plt.tight_layout(); plt.show()

# 3. Optimal Hourly Power Flow Summary
plt.figure(figsize=(14,6))
for col in ["Load_kW", "Solar_kW", "Grid_Import_kW", "Grid_Export_kW", "Battery_Charge_kW", "Battery_Discharge_kW"]:
    plt.plot(df2["Timestamp"], df2[col], label=col)
plt.title("Optimal Hourly Power Flow")
plt.xlabel("Time"); plt.ylabel("Power (kW)"); plt.legend(); plt.xticks(rotation=45)
plt.tight_layout(); plt.show()

```

```python
random_date = np.random.choice(df2["Timestamp"].dt.date.unique())
one_day2 = df2[df2["Timestamp"].dt.date == random_date]

plt.figure(figsize=(10,5))
plt.plot(one_day2["Timestamp"], one_day2["Load_kW"], label="Load")
plt.plot(one_day2["Timestamp"], one_day2["Solar_utilized_kW"], label="Solar Utilized")
plt.title(f"Solar Utilization vs Load – {random_date}")
plt.xlabel("Time"); plt.ylabel("Power (kW)"); plt.legend()
plt.xticks(rotation=45); plt.tight_layout(); plt.show()

# 4. Solar Utilization vs Load
plt.figure(figsize=(12,5))
plt.plot(df2["Timestamp"], df2["Load_kW"], label="Load")
plt.plot(df2["Timestamp"], df2["Solar_utilized_kW"], label="Solar Utilized")
plt.title("Solar Utilization vs Load")
plt.xlabel("Time"); plt.ylabel("Power (kW)"); plt.legend(); plt.xticks(rotation=45)
plt.tight_layout(); plt.show()
```

```python
df3 = pd.read_csv("comparison_schedule.csv")
df3["Timestamp"] = pd.to_datetime(df3["Timestamp"])

random_date = np.random.choice(df3["Timestamp"].dt.date.unique())
one_day3 = df3[df3["Timestamp"].dt.date == random_date]

plt.figure(figsize=(10,5))
plt.plot(one_day3["Timestamp"], one_day3["Heur_Import"], label="Heuristic Import", linestyle='--')
plt.plot(one_day3["Timestamp"], one_day3["LP_Import"], label="LP Import")
plt.title(f"Grid Import – Heuristic vs LP – {random_date}")
plt.xlabel("Time"); plt.ylabel("Power (kW)"); plt.legend()
plt.xticks(rotation=45); plt.tight_layout(); plt.show()

# 5. Heuristic vs LP - Grid Import
plt.figure(figsize=(12,5))
plt.plot(df3["Timestamp"], df3["Heur_Import"], label="Heuristic Import", linestyle='--')
plt.plot(df3["Timestamp"], df3["LP_Import"], label="LP Optimized Import")
plt.title("Grid Import: Heuristic vs LP")
plt.xlabel("Time")
plt.ylabel("Power (kW)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


```
